# Data Cleaning and Machine Learning Preparation

This section creates a cleaned version of the Titanic dataset for machine learning.
The original DataFrame, `df`, is kept unchanged for EDA.

```python
df_ml.isnull().sum().sort_values(ascending=False   )
```

```python
df_ml['Age'].isnull().sum()
```

```python
df_ml['Embarked'].isnull().sum()
```

```python
df_ml = df_ml.drop(columns = 'Cabin')
```

```python
df_ml.isnull().sum().sort_values(ascending=False   )
```

## Data Cleaning Summary

- Created a separate copy of the original dataset named `df_ml`.
- Filled missing `Age` values using the median age.
- Filled missing `Embarked` values using the most common embarkation port.
- Dropped the `Cabin` column because it contained a large proportion of missing values.
- Kept the original `df` unchanged for exploratory analysis.

The dataset is split into training and testing sets before preprocessing.

```python
model_features = [
    "Pclass",
    "Sex",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "Embarked",
    "FamilySize",
    "IsAlone"
]
```

## Split Summary

- Input features were stored in `X_raw`, and survival outcomes were stored in `y`.
- The dataset was split into 80% training data and 20% testing data.
- `random_state=42` was used to make the split reproducible.
- `stratify=y` was used to preserve a similar survival distribution in the training and testing sets.
- Preprocessing will be fitted on the training data only to prevent data leakage.

# Professional Data Preprocessing Pipeline

This pipeline handles missing values and categorical encoding. It is fitted using training data only to prevent data leakage.


```python
numerical_features = [
    "Pclass",
    "Age",
    "SibSp",
    "Parch",
    "Fare",
    "FamilySize",
    "IsAlone"
]

categorical_features = [
    "Sex",
    "Embarked"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore"))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
X_train_prepared = preprocessor.fit_transform(X_train_raw)
```

```python
X_test_prepared = preprocessor.transform(X_test_raw)
```

```python
X_train_prepared.shape
```

```python
X_test_prepared.shape 
```

```python
feature_names = preprocessor.get_feature_names_out()

feature_names
```

## Preprocessing Summary

- Numerical features were processed by filling missing values with the median calculated from the training data.
- Categorical features were processed by filling missing values with the most frequent training value.
- Categorical values were converted into numerical features using one-hot encoding.
- The preprocessing rules were fitted only on training data and then applied to testing data.
- This workflow prevents data leakage.

# Logistic Regression Model

This section builds a machine learning pipeline that preprocesses the data and trains a Logistic Regression classifier to predict Titanic passenger survival.

```python
from sklearn.linear_model import LogisticRegression
```

```python
logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(max_iter=1000, random_state=42))
    ]
)
```

```python
logistic_model.fit(X_train_raw,y_train)
```

```python
y_pred = logistic_model.predict(X_test_raw)
```

```python
y_pred[:10]
```

## Model Training Summary

- A Logistic Regression classifier was used as the first baseline model.
- The preprocessing pipeline and classification model were combined into one workflow.
- The model was trained using raw training features and the corresponding survival outcomes.
- The preprocessing pipeline automatically handled missing values and categorical encoding.
- Predictions were generated using unseen test data.

# Model Evaluation

This section evaluates the Logistic Regression model using unseen test data. Predictions are compared with the actual survival outcomes.

```python
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay
)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print(f"Test Accuracy: {accuracy:.2%}")
```

```python
cm = confusion_matrix(y_test, y_pred)
cm
```

```python
display = ConfusionMatrixDisplay(
    confusion_matrix = cm,
    display_labels = ["Not Survived", "Survived"]
    )
display.plot(cmap="Blues")
plt.title("Logistic Regression Confusion Matrix")
plt.show()
```

### Confusion Matrix Interpretation

- The model correctly predicted 97 passengers who did not survive.
- The model correctly predicted 47 passengers who survived.
- The model incorrectly predicted survival for 13 passengers who did not survive.
- The model incorrectly predicted non-survival for 22 passengers who actually survived.
- Most correct predictions appear on the main diagonal of the confusion matrix.

```python
print(
    classification_report(
        y_test,
        y_pred,
        target_names=["Did Not Survive", "Survived"]
    )
)
```

## Logistic Regression Results

- The Logistic Regression model achieved a test accuracy of 80.45%.
- The confusion matrix shows how many survival and non-survival predictions were correct and incorrect.
- The model performed worse at predicting survivors than non-survivors based on precision and recall.
- Accuracy alone does not fully describe performance because the dataset has more non-survivors than survivors.
- Logistic Regression provides a useful baseline model for comparison with more complex models later.

# Random Forest Hyperparameter Tuning

This section uses cross-validation and GridSearchCV to test different Random Forest settings and select the best model using training data only.

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    "classifier__n_estimators": [200, 400],
    "classifier__max_depth": [None, 5, 10],
    "classifier__min_samples_split": [2, 5],
    "classifier__min_samples_leaf": [1, 2]
}
```

```python
rf_grid_search = GridSearchCV(
    estimator=random_forest_model,
    param_grid=param_grid,
    scoring="f1",
    cv=5,
    n_jobs=-1,
    verbose=1
)
```

```python
rf_grid_search.fit(X_train_raw, y_train)
```

```python
rf_grid_search.best_params_
```

```python
print(f"Best Cross-Validation F1-score: {rf_grid_search.best_score_:.2%}")
```

```python
best_rf_model = rf_grid_search.best_estimator_
```

```python
best_rf_pred = best_rf_model.predict(X_test_raw)
```

## Hyperparameter Tuning Summary

- GridSearchCV was used to test multiple Random Forest hyperparameter combinations.
- Five-fold cross-validation was applied to the training data.
- F1-score was used as the scoring metric to balance precision and recall for survivor predictions.
- The best parameter combination was selected using training cross-validation results only.
- The final tuned model was then used to predict the unseen test data.

# Final Model Evaluation
